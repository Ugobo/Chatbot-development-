## Part 2 — Your first chatbot

This is about as simple as a chatbot gets: it looks for a keyword in what you type, and replies with a
fixed response. No AI, no training — just `if` statements. This is a **rule-based chatbot**.

```python
def tiny_bot(message):
    message = message.lower()
    if "hello" in message or "hi" in message:
        return "Hello! I'm a tiny rule-based bot."
    elif "help" in message:
        return "I can only respond to a few words right now: hello, help, bye."
    elif "bye" in message:
        return "Goodbye!"
    else:
        return "Sorry, I don't understand that yet."

# Try it:
print(tiny_bot("Hello there"))
print(tiny_bot("Can you help me?"))
print(tiny_bot("What's the weather today?"))
print(tiny_bot("bye"))
```

## Part 3 — More advanced - Conversational Tiny Bot (3-Step Flow)


```python
# a 3 step
def tiny_bot():
    print("Bot: Hello! I'm a tiny rule-based bot. What is your name?")
    step = 1

    while True:
        user_input = input("You: ").lower()

        # Handle early exit
        if "bye" in user_input:
            print("Bot: Goodbye!")
            break

        # Step 1: Getting the user's name
        if step == 1:
            name = user_input.capitalize()
            print(f"Bot: Nice to meet you, {name}! How can I help you today? (Ask for 'help' if unsure)")
            step = 2

        # Step 2: Addressing their request
        elif step == 2:
            if "help" in user_input:
                print("Bot: I can answer general questions. Are you feeling happy or sad today?")
                step = 3
            elif "weather" in user_input:
                print("Bot: I don't have a weather app, but I hope it is sunny! Are you feeling happy or sad today?")
                step = 3
            else:
                print("Bot: Interesting! Tell me, are you feeling happy or sad today?")
                step = 3

        # Step 3: Emotional check-in and wrap-up
        elif step == 3:
            if "happy" in user_input:
                print("Bot: That's amazing to hear! Have a wonderful rest of your day. Goodbye!")
            elif "sad" in user_input:
                print("Bot: I'm sorry to hear that. I hope your day gets better! Goodbye!")
            else:
                print("Bot: Thanks for sharing! Have a great day. Goodbye!")
            break # End the conversation after 3 steps

# Run the chatbot
tiny_bot()
```
